# Evaluation and Comparison of Recommendation Models

## 1. Introduction

This notebook will compare the recommendation approaches developed in the Online Course Recommender System project.

The project includes two content-based recommenders:

- A TF-IDF + cosine similarity recommender
- A KNN-based content recommender

The purpose of this evaluation notebook is to compare how these approaches behave, what kinds of recommendations they produce, and where each method is strongest. This comparison will help determine which model is more suitable as a final recommendation approach for the project.

## 2. Evaluation Methodology

The dataset used in this project does not contain user interaction data such as ratings, clicks, watch history, enrollments, likes, or purchases.

Because of this, traditional supervised classification metrics are not directly applicable:

- **Accuracy** requires known correct and incorrect predictions.
- **Precision** requires a ground-truth set of relevant recommendations.
- **Recall** requires knowing all relevant items for each user or query.
- **F1 Score** depends on precision and recall, so it also requires ground-truth relevance labels.

Since the project is based on course metadata rather than user behavior, the evaluation will focus on qualitative and comparative analysis instead.

The planned evaluation areas are:

- **Recommendation relevance:** whether recommended courses appear related to the input course.
- **Recommendation diversity:** whether the model recommends useful alternatives instead of nearly identical courses only.
- **Runtime performance:** how quickly each method generates recommendations.
- **Recommendation overlap:** how many recommendations are shared between TF-IDF and KNN for the same input course.
- **Ablation study:** how recommendation behavior changes when different text features are included.

## 6. Runtime Comparison

This section compares the execution time of the TF-IDF and KNN recommendation approaches.

Runtime comparison is useful because recommendation systems must often generate results quickly, especially when users interact with an application in real time.

The same test course was used for both models:

**Test course:** `web programming with python`

Recommendation generation time was measured once and also averaged across 100 repeated recommendation calls.

| Model | Setup Time (seconds) | Single Recommendation Time (seconds) | Average Recommendation Time over 100 Runs (seconds) |
|---|---:|---:|---:|
| TF-IDF + Cosine Similarity | 0.077497 | 0.001640 | 0.001245 |
| KNN Recommender | 0.058735 | 0.029597 | 0.028309 |

### Runtime Discussion

The TF-IDF recommender is faster during recommendation generation. Its average recommendation time was about `0.001245` seconds, while the KNN recommender averaged about `0.028309` seconds.

The KNN setup time was slightly faster in this run, but the difference is small. The more important difference is recommendation generation time, where TF-IDF is faster because it retrieves similarity scores from a precomputed cosine similarity matrix. KNN performs a nearest-neighbor search over the feature matrix for the selected course.

For this dataset of 3,672 courses, both methods are fast enough for project use. The runtime difference is measurable, but not large enough to be a practical problem at this dataset size. For a much larger course catalog, runtime and memory efficiency would become more important.

The best feature combination for the final recommender is **title + subject + level**. This combination should be used because it preserves the strong topic signal from the title, adds subject-level category information, and includes difficulty level for more specific course matching. This is also consistent with the `combined_features` column created in preprocessing and used by the TF-IDF recommender.

## 8. Comparative Discussion

This section compares the strengths and weaknesses of the TF-IDF and KNN approaches using the recommendation examples, overlap analysis, runtime comparison, and ablation study.

### TF-IDF + Cosine Similarity

**Strengths:**

- Produces highly focused recommendations based on shared course terms.
- Performs especially well when the input course contains distinctive keywords such as `python`, `investment banking`, or `web development`.
- Easy to interpret because high-scoring recommendations usually share visible text with the input course.

**Weaknesses:**

- Can recommend courses that are very close in wording, which may reduce diversity.
- Depends heavily on the available text in `combined_features`.
- Does not use numeric course context such as subscribers, reviews, price, or duration.

In the generated examples, TF-IDF gave very direct matches. For the web development input course, it recommended several courses containing `from scratch`, `web development`, `developer`, JavaScript, and PHP. This makes the results easy to explain, but also shows that TF-IDF tends to stay close to the input wording.

TF-IDF was also faster during recommendation generation. For the test course `web programming with python`, its average recommendation time over 100 runs was `0.001245` seconds, compared with `0.028309` seconds for KNN. This makes TF-IDF a strong choice when simplicity, interpretability, and fast recommendation lookup are priorities.

### KNN Recommender

**Strengths:**

- Can combine text similarity with numeric course metadata.
- Produces relevant results while sometimes adding more variety than TF-IDF.
- Useful for comparing courses in a richer feature space than text alone.

**Weaknesses:**

- Recommendation behavior depends on feature scaling and numeric feature weighting.
- Results may be less directly interpretable than TF-IDF because both text and numeric attributes influence distance.
- Broader recommendations may sometimes be less exact topical matches.

In the generated examples, KNN remained relevant across Python, Business/Finance, and Web Development. For the web development input course, KNN recommended courses about BackboneJS, JavaScript, WCF, web APIs, and becoming a web developer. These are related alternatives, but they are more varied than the TF-IDF list.

KNN had slightly faster setup time in the runtime test, but recommendation generation was slower than TF-IDF for this implementation. This difference is not a practical problem for 3,672 courses, but it would matter more for a larger course catalog.

### Overall Comparison

TF-IDF is stronger as a clear and interpretable baseline because its recommendations are closely tied to textual similarity. KNN is more flexible because it combines text and numeric metadata, but this also makes its results more dependent on feature engineering choices.

The overlap analysis supports this comparison. The two models agree on 60% of recommendations for the Python and Business/Finance examples, but only 40% for the Web Development example. This suggests that both methods identify obvious topic matches, while KNN may introduce more diversity when the feature space supports broader alternatives.

The ablation study shows that `course_title` alone already performs well for the Python example because the title contains strong keywords. Adding `subject` reinforces category relevance, and adding `level` provides useful difficulty context. Therefore, the full `title + subject + level` feature combination remains the best choice for the final content representation.

## 9. Conclusions

This notebook compared the TF-IDF and KNN content-based recommendation models using recommendation relevance, overlap, runtime performance, and feature ablation.

Because the dataset does not include user interaction labels, traditional supervised metrics such as Accuracy, Precision, Recall, and F1 Score were not directly applicable. Instead, the evaluation focused on whether recommendations were relevant, how much the two models agreed, how quickly they generated recommendations, and how feature choices affected results.

The TF-IDF recommender produced highly focused and interpretable recommendations. It was especially strong for courses with distinctive terms such as `python`, `investment banking`, and `web development`. It also generated recommendations faster than KNN in the runtime comparison.

The KNN recommender also produced relevant recommendations, but its results were sometimes more varied because it used both text features and numeric course metadata. This made KNN useful for generating broader alternatives, although its behavior depends more on feature scaling and weighting choices.

The overlap analysis showed that both models agreed on 60% of recommendations for the Python and Business/Finance examples, and 40% for the Web Development example. This indicates that both models capture core topic similarity, while KNN can introduce more diverse recommendations in some cases.

The ablation study showed that `title + subject + level` is the best final text representation. This combination keeps the strong topic signal from the course title, adds subject category information, and includes level information for learner-specific relevance.

Overall, TF-IDF is the best baseline model for this project because it is simple, fast, interpretable, and consistently relevant. KNN is a useful comparison model because it demonstrates how recommendations change when text features are combined with numeric metadata.
